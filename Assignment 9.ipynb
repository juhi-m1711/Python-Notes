## Assignment 9 – Data Visualization

1. Import Seaborn and Matplotlib. Load the tips dataset and display first 5 rows.

```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

tips.head()
```

2. Plot a Histogram of total_bill.

```python
plt.figure(figsize = (12,6))
sns.histplot(data = tips, x = "total_bill", color = "pink")
plt.title("Histogram of total_bill", fontweight = "bold")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', linewidth=0.6, alpha=0.4)
plt.show()
```

3. Plot a Boxplot for tip.

```python
plt.figure(figsize = (12,6))
sns.boxplot(data = tips, y = "tip", color = "#90CAF9")
plt.title("Boxplot for Tips", fontweight = "bold")
plt.grid(axis='y', linestyle='--', linewidth=0.6, alpha=0.4)
plt.ylabel("Tips")
plt.show()
```

4. Plot a Bar Plot for average tip by sex.

```python
plt.figure(figsize =(15,5))
sns.barplot(data = tips, x = "sex", y = "tip", color = "#ABDEEC")
plt.title("Average tip by Sex", fontweight = "bold")
plt.xlabel("Sex")
plt.ylabel("Average Tip")
plt.show()
```

5. Plot a Pie Chart showing proportion of smoker values.

```python
smoker_counts = tips["smoker"].value_counts()
colors = sns.color_palette("pastel")

plt.figure()
plt.pie(smoker_counts, labels=smoker_counts.index, autopct="%1.1f%%", colors = colors)
plt.title("Smoker Distribution", fontweight = "bold")
plt.legend(frameon=False, fontsize=11)
plt.show()
```

6. Create a Scatterplot between total_bill and tip.

```python
plt.figure(figsize =(15,6))
sns.scatterplot(data = tips, x = "total_bill", y = "tip")
plt.title("Total Bill v/s Tip", fontweight = "bold")
plt.xlabel("Total Bill")
plt.ylabel("Tips")
plt.show()
```

7. Create a Line plot showing total bill over the dataset index.

```python
plt.figure(figsize = (20,6))
plt.plot(tips.index, tips["total_bill"])
plt.title("Total Bill over Index", fontweight ="bold")
plt.xlabel("Index")
plt.ylabel("Total Bill")
plt.show()
```

8. Plot a Countplot of day column.

```python
plt.figure(figsize=(12,6))
sns.countplot(data = tips, x = "day", color ="#CCE2C6")
plt.title("Count of Days", fontweight ="bold")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```

9. Plot a Violinplot of tip by day.

```python
plt.figure(figsize=(12,6))
sns.violinplot(data = tips, x = "day", y = "tip")
plt.title("Tip Distribution by Day", fontweight = "bold")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.show()
```

10. Create a Heatmap of correlation between numerical columns in tips.

```python
plt.figure()
corr = tips.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap", fontweight = "bold")
plt.show()
```

11. Use Pairplot to visualize all numerical columns.

```python
plt.figure()
sns.pairplot(tips)
plt.show()
```

12. Plot a Stripplot showing tip distribution by smoker.

```python
plt.figure()
sns.stripplot(data=tips, x="smoker", y="tip")
plt.title("Tip by Smoker", fontweight ="bold")
plt.xlabel("Smoker")
plt.ylabel("Tip")
plt.show()
```

13. Plot a Swarmplot showing tip by day.

```python
plt.figure()
sns.swarmplot(data = tips, x = "day", y = "tip")
plt.title("Tip by day", fontweight = "bold")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.show()
```

14. Create a LMplot between total_bill and tip with regression line.

```python
sns.lmplot(data = tips, x = "total_bill", y = "tip")
plt.title("Regression: Total bill v/s Tip", fontweight = "bold")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

15. Create a Relplot to visualize total_bill vs tip by time using hue.

```python
sns.relplot(data = tips, x = "total_bill", y = "tip", hue = "time")
plt.title("Total bill v/s Tip by time", fontweight = "bold")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```
